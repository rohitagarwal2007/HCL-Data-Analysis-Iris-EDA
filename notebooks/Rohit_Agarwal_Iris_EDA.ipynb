```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('/content/iris_clean.csv')

# Display first and last 5 rows
print("Head:\n", df.head())
print("\nTail:\n", df.tail())

# Display dataset shape (rows and columns)
print("\nShape:", df.shape)

# Display column info and data types
print("\nInfo:")
df.info()

# Display statistical summary
print("\nDescribe:\n", df.describe())
```

```python
# Count records per species using value_counts()
species_count = df['species'].value_counts()
print("Record count for each species:\n", species_count)
```

```python
# Convert columns to NumPy arrays for calculations
sepal_len_arr = np.array(df['sepal_length'])
petal_len_arr = np.array(df['petal_length'])

print("--- Sepal Length ---")
print("Mean:", np.mean(sepal_len_arr))
print("Median:", np.median(sepal_len_arr))
print("Min:", np.min(sepal_len_arr))
print("Max:", np.max(sepal_len_arr))

print("\n--- Petal Length ---")
print("Mean:", np.mean(petal_len_arr))
print("Median:", np.median(petal_len_arr))
print("Min:", np.min(petal_len_arr))
print("Max:", np.max(petal_len_arr))
```

```python
# Group by species to calculate average measurements
avg_measurements = df.groupby('species').mean()
print("Average measurements by species:\n", avg_measurements)

# Identify species with the highest average petal length and petal width
max_petal_species = avg_measurements[['petal_length', 'petal_width']].idxmax()
print("\nSpecies with highest average petal dimensions:\n", max_petal_species)
```

```python
# Sort by petal_length in descending order and get top 5
top_5_petal = df.sort_values(by='petal_length', ascending=False).head(5)
print("Top 5 flowers with largest petal length:\n", top_5_petal)
```

```python
# Check for missing values
print("Missing Values:\n", df.isnull().sum())

# Check total number of duplicate rows
print("\nTotal Duplicate Rows:", df.duplicated().sum())

# Filter and display the duplicate rows
print("\nDuplicate Rows:\n", df[df.duplicated()])
```

```python
# Filter records where petal_length is greater than 6.0 cm
large_petals = df[df['petal_length'] > 6.0]
print("Flowers with petal length > 6.0 cm:\n", large_petals)
```

```python
# Chart 1: Species-Count Bar Chart
plt.figure(figsize=(6, 4))
df['species'].value_counts().plot(kind='bar', color=['skyblue', 'lightgreen', 'salmon'])
plt.title('Count of Each Iris Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()

# Chart 2: Petal-Length Histogram
plt.figure(figsize=(6, 4))
plt.hist(df['petal_length'], bins=15, color='purple', edgecolor='black')
plt.title('Distribution of Petal Length')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Frequency')
plt.show()
```
